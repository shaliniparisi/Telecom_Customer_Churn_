# Telecom Churn Prediction

## 1. Imports and Paths

```python
from pathlib import Path
import json, joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, RandomizedSearchCV, StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (classification_report, confusion_matrix, accuracy_score,
    precision_score, recall_score, f1_score, average_precision_score, roc_auc_score)
from xgboost import XGBClassifier

RANDOM_STATE = 42
THRESHOLD = 0.30
ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
DATA_DIR, MODEL_DIR = ROOT / "data", ROOT / "models"
MODEL_DIR.mkdir(exist_ok=True)
sns.set_theme(style="whitegrid")
```

## 2. Load and Validate data

```python
train_df = pd.read_csv(DATA_DIR / "churn-bigml-80.csv")
test_df = pd.read_csv(DATA_DIR / "churn-bigml-20.csv")

def normalize_churn(s):
    if s.isna().any():
        raise ValueError("Churn contains missing values.")
    if s.dtype == bool:
        return s
    mapping = {"true": True, "false": False, "yes": True, "no": False, "1": True, "0": False}
    out = s.astype(str).str.strip().str.lower().map(mapping)
    if out.isna().any():
        raise ValueError("Unexpected Churn values.")
    return out.astype(bool)

for frame in (train_df, test_df):
    for col in frame.select_dtypes(include=["str","object"]).columns:
        frame[col] = frame[col].str.strip()
    frame["Churn"] = normalize_churn(frame["Churn"])

print("Train shape:", train_df.shape, "| Test shape:", test_df.shape)
print("Train missing:", train_df.isna().sum().sum(), "| Test missing:", test_df.isna().sum().sum())
print("Train duplicates:", train_df.duplicated().sum(), "| Test duplicates:", test_df.duplicated().sum())
print("Train churn counts:\n", train_df["Churn"].value_counts())
print("Test churn counts:\n", test_df["Churn"].value_counts())
display(train_df.head())
```

## 3. Prepare features

```python
charge_cols = ["Total day charge", "Total eve charge", "Total night charge", "Total intl charge"]
for c in charge_cols:
    if c not in train_df.columns or c not in test_df.columns:
        raise ValueError(f"Expected column not found: {c}")

X_all = train_df.drop(columns=["Churn"] + charge_cols)
y_all = train_df["Churn"].astype(bool)
X_test = test_df.drop(columns=["Churn"] + charge_cols)[X_all.columns]
y_test = test_df["Churn"].astype(bool)

X_train, X_val, y_train, y_val = train_test_split(
    X_all, y_all, test_size=0.20, random_state=RANDOM_STATE, stratify=y_all
)
print("Train:", X_train.shape, "Validation:", X_val.shape, "Test:", X_test.shape)
print("Test columns match:", X_test.columns.equals(X_train.columns))
```

## 4. Preprocessing and Baseline models

```python
categorical_cols = ["State", "Area code", "International plan", "Voice mail plan"]
numerical_cols = [c for c in X_train.columns if c not in categorical_cols]

preprocessor = ColumnTransformer([
    ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
    ("numerical", "passthrough", numerical_cols)
])
logistic_preprocessor = ColumnTransformer([
    ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
    ("numerical", StandardScaler(), numerical_cols)
])

class_weight_ratio = (y_train == False).sum() / (y_train == True).sum()

models = {
    "Balanced Logistic Regression": Pipeline([
        ("preprocessor", logistic_preprocessor),
        ("model", LogisticRegression(class_weight="balanced", max_iter=3000, random_state=RANDOM_STATE))
    ]),
    "Random Forest": Pipeline([
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(n_estimators=200, class_weight="balanced",
                                         random_state=RANDOM_STATE, n_jobs=-1))
    ]),
    "XGBoost": Pipeline([
        ("preprocessor", preprocessor),
        ("model", XGBClassifier(n_estimators=200, max_depth=4, learning_rate=0.05,
                                scale_pos_weight=class_weight_ratio, eval_metric="logloss",
                                random_state=RANDOM_STATE, n_jobs=-1))
    ])
}

rows = []
for name, model in models.items():
    model.fit(X_train, y_train)
    prob = model.predict_proba(X_val)[:, 1]
    pred = model.predict(X_val)
    rows.append({"Model": name, "Accuracy": accuracy_score(y_val, pred),
                 "Churn precision": precision_score(y_val, pred, zero_division=0),
                 "Churn recall": recall_score(y_val, pred, zero_division=0),
                 "Churn F1": f1_score(y_val, pred, zero_division=0),
                 "PR-AUC": average_precision_score(y_val, prob),
                 "ROC-AUC": roc_auc_score(y_val, prob)})
display(pd.DataFrame(rows).sort_values("PR-AUC", ascending=False).round(4))
```

## 5. Hyperparameter tuning

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

xgb_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", XGBClassifier(scale_pos_weight=class_weight_ratio, eval_metric="logloss",
                            random_state=RANDOM_STATE, n_jobs=-1))
])
xgb_params = {
    "model__n_estimators": [100, 200, 300, 400],
    "model__max_depth": [2, 3, 4, 5, 6],
    "model__learning_rate": [0.01, 0.03, 0.05, 0.1],
    "model__subsample": [0.8, 0.9, 1.0],
    "model__colsample_bytree": [0.8, 0.9, 1.0]
}
xgb_search = RandomizedSearchCV(xgb_pipeline, xgb_params, n_iter=20, scoring="average_precision",
                                cv=cv, random_state=RANDOM_STATE, n_jobs=-1, verbose=1)
xgb_search.fit(X_train, y_train)
best_xgb_model = xgb_search.best_estimator_
print("XGBoost best parameters:", xgb_search.best_params_)
print("XGBoost best CV PR-AUC:", round(xgb_search.best_score_, 4))

rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(class_weight="balanced", random_state=RANDOM_STATE, n_jobs=-1))
])
rf_params = {
    "model__n_estimators": [100, 200, 300, 400],
    "model__max_depth": [None, 10, 20, 30],
    "model__min_samples_split": [2, 5, 10],
    "model__min_samples_leaf": [1, 2, 4],
    "model__max_features": ["sqrt", "log2"]
}
rf_search = RandomizedSearchCV(rf_pipeline, rf_params, n_iter=20, scoring="average_precision",
                               cv=cv, random_state=RANDOM_STATE, n_jobs=-1, verbose=1)
rf_search.fit(X_train, y_train)
best_rf_model = rf_search.best_estimator_
print("Random Forest best parameters:", rf_search.best_params_)
print("Random Forest best CV PR-AUC:", round(rf_search.best_score_, 4))
```

## 6. Threshold selection

```python
xgb_val_prob = best_xgb_model.predict_proba(X_val)[:, 1]
threshold_rows = []
for threshold in [0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60, 0.65, 0.70]:
    pred = (xgb_val_prob >= threshold).astype(int)
    threshold_rows.append({"Threshold": threshold,
        "Precision": precision_score(y_val, pred, zero_division=0),
        "Recall": recall_score(y_val, pred, zero_division=0),
        "F1": f1_score(y_val, pred, zero_division=0)})
display(pd.DataFrame(threshold_rows).round(4))

THRESHOLD = 0.30
comparison = []
for name, model in [("Tuned Random Forest", rf_search.best_estimator_),
                    ("Tuned XGBoost", best_xgb_model)]:
    prob = model.predict_proba(X_val)[:, 1]
    pred = (prob >= THRESHOLD).astype(int)
    comparison.append({"Model": name, "Threshold": THRESHOLD,
        "Accuracy": accuracy_score(y_val, pred),
        "Churn precision": precision_score(y_val, pred, zero_division=0),
        "Churn recall": recall_score(y_val, pred, zero_division=0),
        "Churn F1": f1_score(y_val, pred, zero_division=0),
        "PR-AUC": average_precision_score(y_val, prob),
        "ROC-AUC": roc_auc_score(y_val, prob)})
display(pd.DataFrame(comparison).round(4))

print("Tuned XGBoost validation report at threshold", THRESHOLD)
print(classification_report(y_val, (xgb_val_prob >= THRESHOLD).astype(int)))
print(confusion_matrix(y_val, (xgb_val_prob >= THRESHOLD).astype(int)))
```

## 7. Evaluate tuned XGBoost on Test data

```python
y_test_prob = best_xgb_model.predict_proba(X_test)[:, 1]
y_test_pred = (y_test_prob >= THRESHOLD).astype(int)
print(classification_report(y_test, y_test_pred))
print("Confusion matrix:\n", confusion_matrix(y_test, y_test_pred))
print("Test accuracy:", round(accuracy_score(y_test, y_test_pred), 4))
print("Test PR-AUC:", round(average_precision_score(y_test, y_test_prob), 4))
print("Test ROC-AUC:", round(roc_auc_score(y_test, y_test_prob), 4))
```

## 8. Feature importance

```python
feature_names = best_xgb_model.named_steps["preprocessor"].get_feature_names_out()
importances = best_xgb_model.named_steps["model"].feature_importances_
importance_df = pd.DataFrame({"Feature": feature_names, "Importance": importances}).sort_values("Importance", ascending=False)
display(importance_df.head(20))
```

## 9. Save model artifacts and verify reload

```python
joblib.dump(best_xgb_model, MODEL_DIR / "telecom_churn_xgboost.pkl")
joblib.dump(THRESHOLD, MODEL_DIR / "churn_threshold.pkl")
with open(MODEL_DIR / "feature_columns.json", "w", encoding="utf-8") as f:
    json.dump(X_train.columns.tolist(), f, indent=2)

loaded_model = joblib.load(MODEL_DIR / "telecom_churn_xgboost.pkl")
loaded_threshold = joblib.load(MODEL_DIR / "churn_threshold.pkl")
print("Threshold:", loaded_threshold)
print("Reload consistency:", np.allclose(
    best_xgb_model.predict_proba(X_test)[:, 1],
    loaded_model.predict_proba(X_test)[:, 1]
))
```

## 10. Reusable prediction function

```python
def predict_churn(customer_data, model=None, threshold=None, feature_columns=None):
    if not isinstance(customer_data, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")
    model = model if model is not None else joblib.load(MODEL_DIR / "telecom_churn_xgboost.pkl")
    threshold = threshold if threshold is not None else joblib.load(MODEL_DIR / "churn_threshold.pkl")
    if feature_columns is None:
        with open(MODEL_DIR / "feature_columns.json", "r", encoding="utf-8") as f:
            feature_columns = json.load(f)
    missing = [c for c in feature_columns if c not in customer_data.columns]
    if missing:
        raise ValueError(f"Missing required columns: {missing}")
    features = customer_data.loc[:, feature_columns].copy()
    if features.isna().any().any():
        raise ValueError("Input contains missing values.")
    probabilities = model.predict_proba(features)[:, 1]
    predictions = (probabilities >= threshold).astype(int)
    return pd.DataFrame({
        "Churn_Probability": probabilities,
        "Predicted_Churn": predictions,
        "Churn_Status": np.where(predictions == 1, "Likely to Churn", "Unlikely to Churn")
    }, index=customer_data.index)

display(predict_churn(X_test.head(5)))
print("Test customers processed:", len(predict_churn(X_test)))
try:
    predict_churn(X_test.head(1).drop(columns=["State"]))
except ValueError as exc:
    print("Missing-column validation passed:", exc)
```
